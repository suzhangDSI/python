```python
# The purpose of the function below and its parameters and their permissile 
# arguments can be found by ?anagram_checker after running it. 

def anagram_checker(word_a, word_b, is_case_sensitive):
    """ 
    Check whether two strings are anagrams of each other, and this is case sensitive. 

    Parameters: 
    word_a and word_b are any two strings provided;
    is_case_sensitive can be either "True" or "False."

    Prints: 
    True or False. 
    """
    if is_case_sensitive == True:
        print(sorted(word_a) == sorted(word_b))
    if is_case_sensitive == False:
        print(sorted(str.lower(word_a)) == sorted(str.lower(word_b)))
```

```python
?anagram_checker
```

```python
anagram_checker("Silent", "listen", False)
```

```python
anagram_checker("Silent", "listen", True)
```

```python
anagram_checker("Silent", "Listen", True)
```
